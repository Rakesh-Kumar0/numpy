```python
# logical operator
# and, or, not

x = np.array([True, False, True])
y = np.array([False, False, True])

print(np.logical_and(x,y))
print(np.logical_or(x,y))
print(np.logical_not(x))
print(np.logical_xor(x,y))
```

## Broadcasting
--Broadcasting is a NumPy feature that allows you to perform operations between arrays of different shapes without manually reshaping or copying the smaller array.

--NumPy compares shapes from right to left and Compare dimensions from right to left. Each pair must be equal or one of them must be 1.

```python
a = np.array([1, 2, 3]) # (3,)
b = np.array([[10], [20]]) # (2,1)
```

```python
print(a.shape)
print(b.shape)
```

```python
a + b
# a --> (2,3) [[1, 2, 3], [1, 2, 3]]
# b --> [[10, 10, 10], [20, 20, 20]]
```

##  statistical functions

```python
data = np.array([1, 2, 3, 4, 5, 6])
```

```python
print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Standard Deviation:", np.std(data))
print("Variance:", np.var(data))
print("Min:", np.min(data))
print("Max:", np.max(data))
```

```python
#np.argmin() and np.argmax() are NumPy functions used to find the position (index) of the minimum and maximum values in an array.
print("Argmin:", np.argmin(data))
print("Argmax:", np.argmax(data))
```

```python
matrix = np.array([[1, 2, 3], 
                   [4, 5, 6]])

print(np.mean(matrix))
print(np.mean(matrix, axis=0))
print(np.mean(matrix, axis=1))
```

```python
data = np.array([1, 3, 5, 7, 9])
print(np.percentile(data, 50)) # range 1 - 100
print(np.quantile(data, 1)) # range 0 - 1 and quantile = 100/percentile​
```

# sort, search, counting functions

```python
arr = np.array([3, 6, 1, 1, 1, 1, 8, 2, 5, 6])
print("Sorted:", np.sort(arr))
print("Sorted indices:", np.argsort(arr))
print("Unique elements:", np.unique(arr))
```

```python
#This line finds the position where 4 should be inserted into the sorted array while keeping the array sorted.
print("Insert 4 position:", np.searchsorted(np.sort(arr), 4))
```

```python
#Even though 0 and 4 aren't present, bincount() creates positions for every integer from 0 up to the maximum value.
print("Count of each element:", np.bincount(arr))
```
