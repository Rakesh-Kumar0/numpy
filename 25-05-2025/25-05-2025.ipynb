## Numpy

```python
import numpy as np
import time
```

```python
# why numpy is faster than plain python

#this is python approch code
a = list(range(100000000))
b = list(range(100000000))

start_t = time.time()
c = [x+y for x,y in zip(a,b)]
end_t = time.time()
time_taken_py = end_t - start_t

print(f"Time taken : {time_taken_py}")
```

```python
# this is numpy approch code

a = list(range(100000000))
b = list(range(100000000))
a = np.array(a)
b = np.array(b)
start_t = time.time()
c = a + b
end_t = time.time()
time_taken_np = end_t - start_t
print(f"Time takesn : {time_taken_np}")
```

```python
#comparision which is better 

time_taken_py/time_taken_np
```


## 1. np.array

```python
array_n1 = np.array([1,2,3])
```

```python
print(array_n1)
print(type(array_n1))
print(type(array_n1[0]))
```

```python
array_n2 = np.array([1,2,3], dtype=np.int8)
print(array_n2)
print(type(array_n2))
print(type(array_n2[0]))
```

```python
array_n2 = np.array([1,2,"3"], dtype=np.int8)
print(array_n2)
print(type(array_n2))
print(type(array_n2[2]))
```

```python
array_n2 = np.array([1,2,"3"] , dtype=np.float16)
print(array_n2)
print(type(array_n2))
print(type(array_n2[2]))
```
