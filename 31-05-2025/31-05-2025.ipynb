## np.eye(n)
-- np.eye() is a NumPy function used to create an identity matrix or a matrix with 1s on a specified diagonal and 0s elsewhere.

```python
np.eye(6,6)
```

```python
np.eye(6,6, dtype=np.int8)
```

```python
np.eye(6,6,4)
```

```python
np.eye(6,6,4,dtype=np.int8)
```

```python
# different row and col
np.eye(3, 4)
```

```python
# by default no. of col is same as row
np.eye(6)
```

```python
np.eye((6,6),4)
```

```python
np.empty((5,5),dtype=np.int8)
```

```python
np.empty((5,5), dtype=int)
```

```python
import time
```

perf_counter() is used to get high-resolution time counter.

```python
start_e = time.perf_counter()
np.empty((5,5))
end_e = time.perf_counter()
total_e = end_e - start_e
print(total_e)
```

```python
start_we = time.perf_counter()
np.ones((5,5))
end_we = time.perf_counter()
total_we = end_we-start_we
print(total_we)
```
